## Example for loading and plotting DRO pickle files

```python
import pickle
import matplotlib.pyplot as plt
import numpy as np
```

Read pickle files

```python
file_dir='orbit_files/'

### reading the pickle file for the 0.86 au orbit
filename='orbit_files/dro06__0_86au.p'
with open(filename,'rb') as f:
    dro = pickle.load(f)

filename='orbit_files/dro09__0_92au.p'
with open(filename,'rb') as f:
    dro2 = pickle.load(f)
```

Make a cartesian plot and a polar plot in HEE coordinates

```python
fig, ax = plt.subplots(1,1,figsize=(5, 5))
ax.plot(dro.x,dro.y)
ax.plot(dro2.x,dro2.y)
ax.set_aspect('equal')

#make a polar plot in HEE
fig, ax = plt.subplots(1,figsize=(5, 5),subplot_kw={'projection': 'polar'},dpi=100)    
ax.scatter(0,0,s=100,c='yellow',alpha=1, edgecolors='black', linewidth=0.3)
ax.plot(np.zeros(11),np.arange(0,1.1,0.1),c='k',lw=2,alpha=1,linestyle='-',zorder=0) #Sun-Earth line
ax.plot(np.deg2rad(np.arange(0,360)),np.zeros(360)+1,lw=1,alpha=1,linestyle='-',c='black', marker=None) #1 au circle
#plot orbit
ax.plot(dro.lon, dro.r,linewidth=1.5, alpha=1.0)
ax.plot(dro2.lon, dro2.r,linewidth=1.5, alpha=1.0)
```
